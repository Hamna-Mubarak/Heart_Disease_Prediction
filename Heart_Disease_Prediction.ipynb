## Introduction

##### **Definition** A binary classification ML task that predicts whether a patient has heart disease (Yes) or does not have heart disease (No) from their clinical attributes (e.g., sex, chest pain type, number of major vessels, thalassemia test result).
##### **Application** It can be used as a decision-support / screening tool to help doctors flag high-risk patients for further tests, and for learning how clinical attributes relate to heart disease.

#### Step 3.2: Pre-process Sample Data
o   Replace the missing values `'?'` with `NaN` and **remove the 6 rows** that contain missing values (4 in `ca`, 2 in `thal`)

o   Convert the output attribute `num` (0–4) into a binary output **HeartDisease**: 0 → **No**, 1–4 → **Yes**

o   Convert the numeric codes of the categorical attributes into readable text values (same style as the Titanic data, e.g. `sex` 1 → **Male**)

````python
"""
Purpose of this section:
-------------------------
This part of the code cleans the raw UCI data so that it has the same form as
the Titanic sample data (readable categorical values, no missing values).
"""

# 1. Replace '?' with NaN, convert columns back to numbers and drop rows with missing values
clean_data = raw_data.replace("?", np.nan)
clean_data["ca"] = pd.to_numeric(clean_data["ca"])
clean_data["thal"] = pd.to_numeric(clean_data["thal"])

print("Rows before removing missing values:", len(clean_data))
clean_data = clean_data.dropna().reset_index(drop=True)
print("Rows after removing missing values :", len(clean_data))

# 2. Convert the output 'num' (0-4) into binary Yes / No
clean_data["HeartDisease"] = np.where(clean_data["num"] > 0, "Yes", "No")

# 3. Convert numeric codes into readable categorical values
clean_data["Sex"] = clean_data["sex"].map({1: "Male", 0: "Female"})
clean_data["ChestPainType"] = clean_data["cp"].map({
    1: "Typical Angina", 2: "Atypical Angina", 3: "Non-Anginal Pain", 4: "Asymptomatic"})
clean_data["MajorVessels"] = clean_data["ca"].map({0: "Zero", 1: "One", 2: "Two", 3: "Three"})
clean_data["Thal"] = clean_data["thal"].map({3: "Normal", 6: "Fixed Defect", 7: "Reversible Defect"})

print("\n\nPre-processed Data (first 10 rows):")
print("===================================\n")
print(clean_data[["sex", "Sex", "cp", "ChestPainType", "ca", "MajorVessels",
                  "thal", "Thal", "num", "HeartDisease"]].head(10))

print("\n\nOutput Distribution after Pre-processing:")
print("=========================================\n")
print(clean_data["HeartDisease"].value_counts())
````

#### Step 4: Feature Extraction

o   **Feature Selection:** As required by the assignment, select only **4 input attributes** (Sex, ChestPainType, MajorVessels, Thal) and the output attribute (HeartDisease), based on the EDA above

````python
"""
Purpose of this section:
-------------------------
This part of the code keeps only the 4 selected input attributes and the
output attribute, in the same format as the Titanic sample data
(4 categorical inputs followed by 1 categorical output).
The result is saved as 'sample-data.csv'.
"""

sample_data = clean_data[["Sex", "ChestPainType", "MajorVessels", "Thal", "HeartDisease"]].copy()

# Save the selected data to a CSV file
sample_data.to_csv(r"sample-data.csv", index=False, header=True)

print("\n\nSample Data with 4 Selected Input Attributes (first 10 rows):")
print("=============================================================\n")
print(sample_data.head(10))
print("\nNumber of Instances in Sample Data:", sample_data["Sex"].count())
print("\nSaved: sample-data.csv")
````

### Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)

#### Step 5.1: Train the Label Encoder

````python
"""
Purpose of this section:
-------------------------
This part of the code sets up and trains Label Encoders.
Label Encoding is used to convert categorical values (like 'Male', 'Female',
'Asymptomatic', 'Normal') into numeric codes that machine learning algorithms can work with.

Steps:
1. Define label categories for each attribute.
2. Initialize a LabelEncoder for each attribute.
3. Train (fit) each LabelEncoder with its respective category values.
"""

# Define label categories for each feature
# These are possible values each column can have in the dataset
sex = pd.DataFrame({"Sex": ["Male", "Female"]})
chest_pain = pd.DataFrame({"ChestPainType": ["Typical Angina", "Atypical Angina", "Non-Anginal Pain", "Asymptomatic"]})
major_vessels = pd.DataFrame({"MajorVessels": ["Zero", "One", "Two", "Three"]})
thal = pd.DataFrame({"Thal": ["Normal", "Fixed Defect", "Reversible Defect"]})
heart_disease = pd.DataFrame({"HeartDisease": ["Yes", "No"]})

# Create LabelEncoder objects for each categorical column
sex_label_encoder = LabelEncoder()
chest_pain_label_encoder = LabelEncoder()
major_vessels_label_encoder = LabelEncoder()
thal_label_encoder = LabelEncoder()
heart_disease_label_encoder = LabelEncoder()

# Train (fit) the encoders on the given categories
# np.ravel() flattens the DataFrame column into a 1D array for the encoder
sex_label_encoder.fit(np.ravel(sex))
chest_pain_label_encoder.fit(np.ravel(chest_pain))
major_vessels_label_encoder.fit(np.ravel(major_vessels))
thal_label_encoder.fit(np.ravel(thal))
heart_disease_label_encoder.fit(np.ravel(heart_disease))

# Show the numeric code assigned to each category
for name, encoder in [("Sex", sex_label_encoder), ("ChestPainType", chest_pain_label_encoder),
                      ("MajorVessels", major_vessels_label_encoder), ("Thal", thal_label_encoder),
                      ("HeartDisease", heart_disease_label_encoder)]:
    print(f"{name:14s}:", {c: int(v) for c, v in zip(encoder.classes_, encoder.transform(encoder.classes_))})
````

#### Step 5.2: Label Encode the Output

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the target/output variable
("HeartDisease") in the dataset. The categorical values ("Yes", "No") are
converted into numerical values (1 and 0) that can be used by
machine learning algorithms.
"""

# Create copies of the dataset (one for encoded output, one to keep original)
sample_data_encoded_output = sample_data.copy()
original_sample_data = sample_data.copy()

# Transform "HeartDisease" (categorical -> numeric) using the trained LabelEncoder
print("\n\nHeartDisease Attribute After Label Encoding (first 10 rows):")
print("============================================================\n")
sample_data["encoded_heart_disease"] = heart_disease_label_encoder.transform(sample_data["HeartDisease"])

# Show original "HeartDisease" values and their numeric encodings
print(sample_data[["HeartDisease", "encoded_heart_disease"]].head(10))

# Replace original categorical output column with encoded values
sample_data_encoded_output[["Sex", "ChestPainType", "MajorVessels", "Thal", "HeartDisease"]] = \
    sample_data[["Sex", "ChestPainType", "MajorVessels", "Thal", "encoded_heart_disease"]]

# Save the encoded dataset to a CSV file
sample_data_encoded_output.to_csv(r"sample-data-encoded-output.csv", index=False, header=True)
print("\nSaved: sample-data-encoded-output.csv")
````

#### Step 5.3: Label Encode the Input

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the input features
("Sex", "ChestPainType", "MajorVessels", "Thal"). Each categorical value
is converted into a numeric value so that machine learning models
can process them.
"""

# Create a copy of the dataset (to keep original values safe)
sample_data_encoded = sample_data_encoded_output.copy()

# Encode each input attribute
sample_data_encoded_output["encoded_sex"] = sex_label_encoder.transform(sample_data_encoded_output["Sex"])
sample_data_encoded_output["encoded_chest_pain"] = chest_pain_label_encoder.transform(sample_data_encoded_output["ChestPainType"])
sample_data_encoded_output["encoded_major_vessels"] = major_vessels_label_encoder.transform(sample_data_encoded_output["MajorVessels"])
sample_data_encoded_output["encoded_thal"] = thal_label_encoder.transform(sample_data_encoded_output["Thal"])

# Replace categorical columns with encoded values in the dataset
sample_data_encoded[["Sex", "ChestPainType", "MajorVessels", "Thal", "HeartDisease"]] = \
    sample_data_encoded_output[["encoded_sex", "encoded_chest_pain", "encoded_major_vessels",
                                "encoded_thal", "HeartDisease"]]
sample_data_encoded = sample_data_encoded.astype(int)

# Print encoded dataset (with numeric values instead of categories)
print("\n\nSample Data after Label Encoding (first 10 rows):")
print("=================================================\n")
print(sample_data_encoded.head(10))

# Save the encoded dataset to a new CSV file
sample_data_encoded.to_csv(r"sample-data-encoded.csv", index=False, header=True)
print("\nSaved: sample-data-encoded.csv")
````

### Step 6: Execute the Training Phase

#### Step 6.1: Splitting Sample Data into Training Data and Testing Data

**Note:** The Titanic notebook used `shuffle=False`. Here `shuffle=True` with `stratify` is used, so that
the training and testing data both contain the same proportion of Yes / No patients
(the UCI file is not guaranteed to be randomly ordered). `random_state=0` keeps the split reproducible.

````python
"""
Purpose of this section:
-------------------------
This part of the code splits the dataset into Training Data and Testing Data.
- Training Data is used to teach the machine learning model.
- Testing Data is used to check how well the model performs on unseen data.
"""

# Split dataset into training (80%) and testing (20%)
# random_state=0 ensures reproducibility (same split every time)
# stratify keeps the same Yes/No ratio in both training and testing data
training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=0,
    shuffle=True,
    stratify=sample_data_encoded["HeartDisease"]
)

# Save the training and testing datasets to CSV files
training_data_encoded.to_csv(r"training-data-encoded.csv", index=False, header=True)
testing_data_encoded.to_csv(r"testing-data-encoded.csv", index=False, header=True)

print("Training Data shape:", training_data_encoded.shape)
print("Testing Data shape:", testing_data_encoded.shape)
````

### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (input vectors) and
the labels (output values) from the training dataset.
"""

# Extract input vectors (features) from training data
input_vector_train = training_data_encoded.iloc[:, :-1]

# Extract output labels (target variable) from training data
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

### Step 6.3: Train the Support Vector Classifier

````python
"""
Purpose of this section:
-------------------------
This part of the code trains a Support Vector Classifier (SVC) on the training data.
The SVC is a type of machine learning model that finds the best boundary
to separate different classes (e.g., heart disease vs. no heart disease).
"""

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

# Initialize the Support Vector Classifier with chosen parameters
svc_model = svm.SVC(gamma="auto", random_state=0)

# Train the model using the training data (features and labels)
svc_model.fit(input_vector_train, np.ravel(output_label_train))

# Print trained model details (parameters and values)
print(svc_model)
````

### Step 6.4: Save the Trained Model

````python
"""
Purpose of this section:
-------------------------
This part of the code saves the trained Support Vector Classifier (SVC)
model to your computer so it can be reused later without retraining.
"""

# Save the trained SVC model to a file named "heart_disease_model.pkl"
with open("heart_disease_model.pkl", "wb") as model_file:
    pickle.dump(svc_model, model_file)
print("Saved: heart_disease_model.pkl")
````

### Step 7: Execute the Testing Phase

#### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (inputs) and the labels (outputs)
from the testing dataset.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

### Step 7.2: Load the Saved Model

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the previously saved Support Vector Classifier (SVC)
model from disk into memory.
"""

with open("heart_disease_model.pkl", "rb") as model_file:
    model = pickle.load(model_file)
print("Model loaded from heart_disease_model.pkl")
````

### Step 7.3: Evaluate the Machine Learning Model

#### Step 7.3.1: Make Predictions with the Trained Model on Testing Data

````python
"""
Purpose of this section:
-------------------------
This part of the code evaluates the trained Support Vector Classifier (SVC)
on the testing data.
"""

model_predictions = model.predict(input_vector_test)

pd.options.mode.chained_assignment = None
testing_data_encoded["Predictions"] = model_predictions
testing_data_encoded.to_csv(r"model-predictions.csv", index=False, header=True)

model_predictions = testing_data_encoded
print("\n\nPredictions Returned by heart_disease_model (first 15 test instances):")
print("====================================================================\n")
print(model_predictions.head(15))
print("\nSaved: model-predictions.csv")
````

### Step 7.4: Calculate the Accuracy Score

````python
"""
Purpose of this section:
-------------------------
This part of the code calculates how accurate the trained model is on the
testing dataset.
"""

model_accuracy_score = accuracy_score(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"]
)

print("\n\nAccuracy Score:")
print("===============\n")
print(round(model_accuracy_score, 2))
````

### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix)

````python
"""
Purpose of this section:
-------------------------
Accuracy alone can be misleading. This section adds the standard classification metrics:
- Precision: of everyone the model predicted as "Heart Disease", how many actually have it?
- Recall: of everyone who actually has heart disease, how many did the model catch?
  (Recall is very important in medicine - a missed patient is a costly error.)
- F1-score: the harmonic mean of precision and recall.
- Confusion Matrix: a table of correct vs. incorrect predictions per class.
"""

print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"],
    target_names=["No Heart Disease (0)", "Heart Disease (1)"]
))

conf_matrix = confusion_matrix(model_predictions["HeartDisease"], model_predictions["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No Heart Disease", "Actual: Heart Disease"],
    columns=["Predicted: No Heart Disease", "Predicted: Heart Disease"]
)
print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues",
            xticklabels=["No Disease", "Disease"],
            yticklabels=["No Disease", "Disease"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - SVC on Heart Disease Testing Data")
plt.tight_layout()
plt.show()
````

### Step 7.6: 10-Fold Cross Validation (Robustness Check)

````python
"""
Purpose of this section:
-------------------------
The testing data has only about 60 patients, so a single accuracy score can
change a lot depending on which patients end up in the test set.
10-fold cross validation trains and tests the same SVC 10 times on different
splits of the whole dataset and reports the average accuracy, which is a
more reliable estimate of how the model will perform on new patients.
"""

all_inputs = sample_data_encoded.iloc[:, :-1]
all_outputs = sample_data_encoded.iloc[:, -1]

cv_scores = cross_val_score(svm.SVC(gamma="auto", random_state=0), all_inputs, all_outputs, cv=10)

print("\n\nAccuracy in each of the 10 folds:")
print("=================================\n")
print(np.round(cv_scores, 2))
print(f"\nMean Accuracy (10-fold): {cv_scores.mean():.2f}  (+/- {cv_scores.std():.2f})")
````
